## Capital estimation (Monte-Carlo)

#### Capital concentration - on PD x LGD

````python
# capital estimation 'monte-carlo' with observed LGDs

capital_montecarlo, _, _, simulated_losses= estimate_capital(portfolio, 
                                                             method="monte-carlo", 
                                                             column_lgd="LossGivenDefault", 
                                                             column_def=settings.COLUMN_TARGET, 
                                                             column_ead=settings.COLUMN_EAD, 
                                                             allocate=True, 
                                                             verbose=False)
heatmap_EL = plot_heatmap(capital_montecarlo, variable="EL" , column_lgd="LossGivenDefault")
heatmap_CA = plot_heatmap(capital_montecarlo, variable="CAP", column_lgd="LossGivenDefault")
fig = plot_loss_distribution(simulated_losses, portfolio[settings.COLUMN_EAD].sum(), show_plot=True)

# CHECK KEY RESULT: output must be as follows
# selected variable:  EL
#      ≥20%
#  0.145875
#  0.259950
#  0.007769
````

#### Capital concentration - on PD

````python
bins = [0, 0.05, 0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.4, 0.5, 0.6, 0.7, 0.8, 0.9, 1.0]

table_EL   = aggregate_el_on_pd(capital_montecarlo, "EL"   , bins)
table_CAP  = aggregate_el_on_pd(capital_montecarlo, "CAP"  , bins)
````

#### Capital reduction

````python
# how do expected loss and economic capital respond 
# to a decreasing PD acceptance threshold ?

pd_thresholds = [1.00, 0.25, 0.20, 0.15, 0.10, 0.08, 0.05]
est_reduction = estimate_capital_reduction(portfolio, 
                                           pd_thresholds, 
                                           column_lgd="LossGivenDefault", 
                                           column_def=settings.COLUMN_TARGET, 
                                           column_ead=settings.COLUMN_EAD,
                                           verbose=True)
show = show_approval_vs_reduction(est_reduction, invert_y=False)
````

#### Monte-Carlo in reverse

````python
# how much should we stress PDs 
# to exceed an economic capital constraint?

pd_stress_factors = np.round(np.arange(1.0, 5.1, 0.1), 1)
capital_stress = estimate_reverse_capital_stress(portfolio, 
                                                 threshold=0.20, 
                                                 stress_factors=pd_stress_factors,
                                                 column_lgd='LossGivenDefault',
                                                 column_def=settings.COLUMN_TARGET,
                                                 column_ead=settings.COLUMN_EAD,
                                                 verbose=False)
````

````python
show_1 = show_pd_stress_vs_capital(capital_stress, show_plot=True, invert_y=False)
show_2 = show_pd_stress_vs_lossdev(capital_stress, show_plot=True, invert_y=False)
````
